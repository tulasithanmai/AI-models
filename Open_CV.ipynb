Bitwise Operations on Binary Images

```python
import numpy as np
```

```python
#Binary Images: Image with only 2 channels black(0) and white(1)
#Bitwise Operators: AND, OR, NOT, XOR
# AND: If both of channels are 1 then it gives 1, if both are 0 then 0, but if both are different like 0,1 or 1,0 it gives us 0.
# OR: If either one of it is 1 it gives 1 and if both are 0 gives 0, both are 1 then 1
#XOR: If both are same then it gives 0 and if both are different it gives 1.
# NOT: if its 0 it converts to 1 and vice versa
img1 = cv2.imread(r"D:\Image Segmentation\tulasi 1\OpenCV practice\yin yang.jfif")
img2 = cv2.imread(r"D:\Image Segmentation\tulasi 1\OpenCV practice\Screenshot 2025-01-10 181034.png")

img1 = cv2.resize(img1,(300,300))
img2 = cv2.resize(img2,(300,300))

# new = cv2.bitwise_and(img1,img2)
# new = cv2.bitwise_or(img1,img2)
# new = cv2.bitwise_xor(img1,img2)
new = cv2.bitwise_not(img1)
new1 = cv2.bitwise_not(img2)

h = np.hstack((img1,new))
h1 = np.hstack((img2,new1))

v = np.vstack((h,h1))

cv2.imshow("tulasi",v)
cv2.waitKey(0)
cv2.destroyAllWindows()

```

### Morphological Operations

 Used to extract image components that are useful in the representation and description of region shape.
 Perfoemd on binary images (Black and white)

 **Types of morphological Operations**
- Erosion: Erodes away the boundaries of fg obj. Mostly fg are white. Used to remove white noises.
- Dilation: Opposite of erosion.Changes black to white channels.
- Opening: Removes the noise. Erosion followed by dilation
- Closing: Dilation followed by erosion.
- Gradient morphological: Dilation - Erosion. It creates like a boundary
- Top hat/White hat: Input image - output image. The most rewritten parts of white is changed to black.
- Black hat: The most rewritten parts of white are kept.

```python
img = cv2.imread(r"D:\Image Segmentation\tulasi 1\OpenCV practice\BW image.png")
img = cv2.resize(img,(200,200))

m = np.ones((5,5),np.uint8) #bigger the matrix thinner the image becomes
m1 = np.ones((10,10),np.uint8)
#Erosion
er = cv2.erode(img,m,iterations = 1)
er1 = cv2.erode(img,m1,iterations = 1)
#Dilate
di = cv2.dilate(img,m,iterations = 1)
#Opening
op = cv2.morphologyEx(img,cv2.MORPH_OPEN,m,iterations = 1)
#Closing
cl = cv2.morphologyEx(img,cv2.MORPH_CLOSE,m, iterations =1)
# Morphological Grdient
gr = cv2.morphologyEx(img,cv2.MORPH_GRADIENT,m, iterations =1)
#Top hat
th = cv2.morphologyEx(img,cv2.MORPH_TOPHAT,m, iterations =1)
#Black Hat
bh = cv2.morphologyEx(img,cv2.MORPH_BLACKHAT,m, iterations =1)

h = np.hstack((img,cl,op,gr,th,bh))
cv2.imshow("tulasi",h)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
