```python

#loading libraries
import pandas as pd
import numpy as np
import statsmodels.api as sm
from sklearn.model_selection import train_test_split
```

```python
credit_df = pd.read_csv("C:/Users/DELL-PC/Desktop/TERM 1/Term 4/IAIML/German Credit Data.csv")
credit_df.head()
credit_df.info()
credit_df.status.value_counts()
```

```python
x_features = list(credit_df.columns)
print(x_features)
x_features.remove('status')
print(x_features)
```

```python
encoded_credit_df = pd.get_dummies(credit_df[x_features],drop_first = True)
print(encoded_credit_df)
```

```python
y = credit_df.status
print(y)
```

```python
x = sm.add_constant(encoded_credit_df)
print(x)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.3, random_state = 42)
logit = sm.Logit(y_train, x_train)
logit_model = logit.fit()
logit_model.summary2()
```

```python
def get_significant_vars(lm):
    var_p_vals_df = pd.DataFrame(lm.pvalues)
    var_p_vals_df['vars'] = var_p_vals_df.index
    var_p_vals_df.columns = ['pvals','vars']
    return list (var_p_vals_df[var_p_vals_df.pvals <= 0.05]['vars'])
```

```python
significant_vars = get_significant_vars(logit_model)
significant_vars

final_logit = sm.Logit(y_train, sm.add_constant(x_train[significant_vars])).fit()
final_logit.summary2(0)
```

```python
# the final model will be P(Y = 1) = 1/(1+e^-(a+b1x1 + b2x2 + ... + b8x8))
# b1x1 = duration * 0.0197 etc..
# if P(Y = 1 ) > 0.5, Y predicted = 1 ,
# else Y predicted = 0.
```

```python
#creating a dataframe of columns actual, predicted prob with final logit model function and aslo appended a new one predicted column.
y_pred_df = pd.DataFrame({"actual": y_test , "predicted_prob": final_logit.predict(sm.add_constant(x_test[significant_vars]))})
y_pred_df['predicted'] = y_pred_df.predicted_prob.map(
    lambda x:1 if x>0.5 else 0) 
print(y_pred_df)
```

```python
import matplotlib.pyplot as plt
from sklearn import metrics

confusion_matrix = metrics.confusion_matrix(y_pred_df.actual , y_pred_df.predicted)
cm_display = metrics.ConfusionMatrixDisplay(confusion_matrix =confusion_matrix , display_labels = ['Good Credit', 'Bad Credit'])
cm_display.plot()
plt.show()
```
